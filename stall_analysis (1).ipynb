## Project Summary
Wanjiku operates a fresh-produce stall at Gikomba Market and has manually tracked two weeks of daily revenue alongside her current inventory levels, reorder thresholds, unit costs, and selling prices. Because she lacks automated reporting tools, she needs a clear, data-backed evaluation of her stall's overall financial performance, week-over-week sales momentum, weekday versus weekend demand patterns, and immediate inventory restocking priorities.

## Top Three Findings
1. **Steady Week-on-Week Revenue Growth:** Total 14-day revenue reached **KES 46,450**, with Week 2 sales (**KES 24,400**) growing by **10.66%** over Week 1 (**KES 22,050**).
2. **Strong Weekend Sales Surge:** Average weekend daily sales (**KES 4,900.00**) are **82.50% higher** than average weekday daily sales (**KES 2,685.00**), peaking at **KES 5,600** on Saturday of Week 2.
3. **Urgent Restock Needed on High-Margin Staples:** **Tomatoes** (18 kg vs. 25 kg reorder level; **33.33%** gross margin) and **Sukuma** (9 kg vs. 15 kg reorder level; **40.00%** gross margin) have fallen below their reorder thresholds and require immediate restocking.
