# Gikomba Market Stall Performance Analysis
**Client:** Wanjiku's Fresh-Produce Stall, Gikomba Market  
**Objective:** Analyze 14-day sales performance, growth trends, day-type patterns, and inventory margins using core Python.


```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: 14-Day Sales Summary & Best/Worst Days
**Approach:** Use Python's built-in `sum()`, `len()`, `max()`, and `min()` functions to compute overall revenue statistics, then locate the positions of the highest and lowest sales using `list.index()` to map each to its corresponding day name and week number (using integer division `// 7 + 1`).


```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
best_week = (best_index // 7) + 1

worst_day = days[worst_index]
worst_week = (worst_index // 7) + 1

print(f"Total 14-Day Sales : KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Best Day           : {best_day} (Week {best_week}) - KES {highest_sales:,}")
print(f"Worst Day          : {worst_day} (Week {worst_week}) - KES {lowest_sales:,}")
```

## Task 2: Weekly Totals & Week-on-Week Growth
**Approach:** Define `week_totals(sales)` using list slicing (`sales[:7]` for Week 1 and `sales[7:]` for Week 2) to return a tuple of weekly totals, unpack those values, and compute the percentage change from Week 1 to Week 2.


```python
def week_totals(sales):
    """Calculate total sales for Week 1 and Week 2 from a 14-day sales list.

    Args:
        sales (list[int | float]): Daily sales figures for 14 days.

    Returns:
        tuple[int | float, int | float]: A tuple containing (week_1_total, week_2_total).
    """
    week_1 = sum(sales[:7])
    week_2 = sum(sales[7:])
    return (week_1, week_2)


w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total       : KES {w1_total:,}")
print(f"Week 2 Total       : KES {w2_total:,}")
print(f"Week-on-Week Growth: {wow_growth:+.2f}%")
```

## Task 3: Daily Performance Classification
**Approach:** Implement `classify_day(amount, average)` to label days exceeding 120% of the average as `"Strong"`, days below 80% of the average as `"Weak"`, and all others as `"Normal"`, then tally the counts across all 14 days using a dictionary.


```python
def classify_day(amount, average):
    """Classify a single day's sales relative to the overall daily average.

    Args:
        amount (int | float): Sales amount for a specific day.
        average (float): Overall average daily sales.

    Returns:
        str: 'Strong' (>20% above average), 'Weak' (>20% below average), or 'Normal'.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    return "Normal"


performance_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for sale in daily_sales:
    category = classify_day(sale, average_sales)
    performance_counts[category] += 1

print("Day Classification Counts:", performance_counts)
```

## Task 4: Weekday vs. Weekend Sales Comparison
**Approach:** Pair each day label with its sales value using `zip(days, daily_sales)`, separate observations into weekday (`Mon`–`Fri`) and weekend (`Sat`–`Sun`) lists, and compute both averages and the percentage uplift on weekends relative to weekdays.


```python
weekday_sales = []
weekend_sales = []

for day, sale in zip(days, daily_sales):
    if day in ("Sat", "Sun"):
        weekend_sales.append(sale)
    else:
        weekday_sales.append(sale)

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend Uplift       : +{weekend_diff_pct:.2f}% higher than weekdays")
```

## Task 5: Inventory Reorder Alerts & Gross Margins
**Approach:** Iterate over `stock.items()` inside `get_reorder_items(inventory)` to collect products whose current stock (`qty`) is less than or equal to their `reorder` threshold, and calculate each product's gross margin percentage using `(price - cost) / price * 100`.


```python
def get_reorder_items(inventory):
    """Identify products whose stock quantity is at or below the reorder level.

    Args:
        inventory (dict): Dictionary mapping product names to stock details.

    Returns:
        list[str]: Product names requiring immediate restocking.
    """
    low_stock = []
    for product, details in inventory.items():
        if details["qty"] <= details["reorder"]:
            low_stock.append(product)
    return low_stock


reorder_items = get_reorder_items(stock)

product_margins = {}
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    product_margins[product] = margin

print("Items to Reorder:", reorder_items)
for product, margin in product_margins.items():
    print(f"{product:<10} Gross Margin: {margin:.2f}%")
```

## Task 6: Formatted Stall Performance Report
**Approach:** Combine all computed metrics into a cleanly aligned executive summary using Python f-strings with comma thousand separators (`:,`) and fixed decimal formatting.


```python
border = "=" * 58
divider = "-" * 58

print(border)
print(f"{'WANJIKU FRESH-PRODUCE STALL PERFORMANCE REPORT':^58}")
print(f"{'Gikomba Market | 14-Day Operational Review':^58}")
print(border)

print("\n1. OVERALL SALES SUMMARY")
print(divider)
print(f"{'Total 14-Day Revenue:':<32} KES {total_sales:>12,}")
print(f"{'Average Daily Revenue:':<32} KES {average_sales:>12,.2f}")
print(f"{'Best Day (Week ' + str(best_week) + ' - ' + best_day + '):':<32} KES {highest_sales:>12,}")
print(f"{'Worst Day (Week ' + str(worst_week) + ' - ' + worst_day + '):':<32} KES {lowest_sales:>12,}")

print("\n2. WEEKLY GROWTH & DAY-TYPE COMPARISON")
print(divider)
print(f"{'Week 1 Sales:':<32} KES {w1_total:>12,}")
print(f"{'Week 2 Sales:':<32} KES {w2_total:>12,}")
print(f"{'Week-on-Week Growth:':<32} {wow_growth:>15.2f}%")
print(f"{'Weekday Average (Mon-Fri):':<32} KES {avg_weekday:>12,.2f}")
print(f"{'Weekend Average (Sat-Sun):':<32} KES {avg_weekend:>12,.2f}")
print(f"{'Weekend vs. Weekday Difference:':<32} +{weekend_diff_pct:>14.2f}%")

print("\n3. DAILY PERFORMANCE BREAKDOWN")
print(divider)
print(f"{'Strong Days (>20% above avg):':<32} {performance_counts['Strong']:>12} days")
print(f"{'Normal Days (within ±20%):':<32} {performance_counts['Normal']:>12} days")
print(f"{'Weak Days (>20% below avg):':<32} {performance_counts['Weak']:>12} days")

print("\n4. INVENTORY STATUS & GROSS MARGINS")
print(divider)
print(f"{'Product':<12} {'Stock':>7} {'Reorder':>8} {'Margin':>9} {'Status':>14}")
for product, info in stock.items():
    status = "RESTOCK NOW" if product in reorder_items else "OK"
    margin = product_margins[product]
    print(
        f"{product:<12} {info['qty']:>5}kg {info['reorder']:>6}kg "
        f"{margin:>8.2f}% {status:>14}"
    )
print(border)
```
